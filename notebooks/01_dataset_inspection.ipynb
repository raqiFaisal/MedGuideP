```python
from google.colab import drive
from pathlib import Path
import sys
import PIL

drive.mount('/content/drive')
project = Path('/content/drive/MyDrive/MedGuide2')
dataset = Path('/content/drive/MyDrive/MedGuideData/detection')
sys.path.insert(0, str(project / 'data_pipeline'))
print('Python', sys.version.split()[0], 'Pillow', PIL.__version__)
```

```python
from prepare_images import inspect, prepare

category_map = {'medicine_box': 'medicine_box'}
samples, report = inspect(dataset / 'images', dataset / 'instances.json',
                          dataset / 'identities.csv', category_map)
```

```python
import json
from PIL import Image, ImageDraw
from IPython.display import display

report_path = dataset / 'inspection_report.json'
report_path.write_text(json.dumps(report, ensure_ascii=False, indent=2), encoding='utf-8')
print({key: value for key, value in report.items() if key != 'excluded'})
for product, path, _, _, labels in samples[:6]:
    with Image.open(path) as original:
        image = original.convert('RGB')
    image.thumbnail((480, 480))
    draw = ImageDraw.Draw(image)
    for label in labels:
        class_id, cx, cy, w, h = map(float, label.split())
        draw.rectangle(((cx-w/2)*image.width, (cy-h/2)*image.height,
                        (cx+w/2)*image.width, (cy+h/2)*image.height), outline='lime', width=3)
    print(product, path.name)
    display(image)
```
